```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
import sys

# --- 1. 한글 폰트 설정 (VM 환경) ---
plt.rc('font', family='Malgun Gothic')

plt.rc('axes', unicode_minus=False)  # 마이너스 부호 깨짐 방지
try:
    font_path = '/usr/share/fonts/truetype/nanum/NanumGothic.ttf'
    font_name = fm.FontProperties(fname=font_path, size=10).get_name()
    plt.rc('font', family=font_name)
except FileNotFoundError:
    print("경고: 나눔고딕 폰트를 찾을 수 없습니다. 기본 폰트로 실행됩니다. (한글이 깨질 수 있음)")
plt.rc('axes', unicode_minus=False) # 마이너스 부호 깨짐 방지
pd.options.display.float_format = '{:,.2f}'.format # Pandas 출력 형식

# --- 2. 데이터 로드 ---
try:
    # 2a. 승객/운행횟수 데이터
    try:
        # [!] DtypeWarning을 무시하지 않고, low_memory=False를 추가 (권장)
        df_data = pd.read_csv('df_merged_final.csv', encoding='cp949', low_memory=False)
    except UnicodeDecodeError:
        df_data = pd.read_csv('df_merged_final.csv', encoding='utf-8', low_memory=False)
    
    # 2b. 정류장 마스터 (행정구 정보 포함)
    try:
        df_master = pd.read_csv('SEOUL_BUS_STOP.csv', encoding='cp949')
    except UnicodeDecodeError:
        df_master = pd.read_csv('SEOUL_BUS_STOP.csv', encoding='utf-8')
        
except FileNotFoundError:
    print("오류: 필요한 CSV 파일('df_merged_final.csv' 또는 'SEOUL_BUS_STOP.csv')을 찾을 수 없습니다.")
    sys.exit() # 중단
except Exception as e:
    print(f"파일 로드 오류: {e}")
    sys.exit() # 중단
    
# --- 3. 컬럼명 정리 및 확인 ---
# 3a. df_data: '총운행\n횟수' -> '총운행횟수'
df_data.rename(columns={'총운행\n횟수': '총운행횟수'}, inplace=True)
# 3b. df_master: 'ARS-ID' -> '버스정류장ARS번호', '행정구명' -> '자치구'
df_master.rename(columns={
    'ARS-ID': '버스정류장ARS번호', 
    '행정구명': '자치구'
}, inplace=True)

# 3c. 필수 컬럼 확인
required_data_cols = ['버스정류장ARS번호', '승차총승객수', '하차총승객수', '총운행횟수', '노선번호']
required_master_cols = ['버스정류장ARS번호', '자치구']

if not all(col in df_data.columns for col in required_data_cols):
    print(f"오류: 'df_merged_final.csv'에 {required_data_cols} 컬럼들이 모두 필요합니다.")
    sys.exit()
if not all(col in df_master.columns for col in required_master_cols):
    print(f"오류: 'SEOUL_BUS_STOP.csv'에 {required_master_cols} 컬럼들이 모두 필요합니다.")
    sys.exit()

# --- [!! 수정된 부분 !!] ---
# 3d. Merge 키('버스정류장ARS번호') 타입을 통일 (dtype mismatch 오류 해결)
# 두 컬럼 모두 문자열(object) 타입으로 변경하여 병합합니다.
try:
    df_data['버스정류장ARS번호'] = df_data['버스정류장ARS번호'].astype(str)
    df_master['버스정류장ARS번호'] = df_master['버스정류장ARS번호'].astype(str)
except Exception as e:
    print(f"Merge 키 타입 변경 중 오류: {e}")
    sys.exit()
# --- [수정 완료] ---


# --- 4. 데이터 병합 (Inner Join) ---
# '정류장ARS번호'를 기준으로 '자치구' 정보를 병합
df_joined = pd.merge(
    df_data,
    df_master[['버스정류장ARS번호', '자치구']], # 마스터에서는 키와 '자치구' 컬럼만
    on='버스정류장ARS번호',
    how='inner' # inner join
)

# --- 5. '자치구'별 총 승하차 인원 계산 ---
df_gu_passengers = df_joined.groupby('자치구')[['승차총승객수', '하차총승객수']].sum()
df_gu_passengers['총승하차_인원'] = df_gu_passengers['승차총승객수'] + df_gu_passengers['하차총승객수']

# --- 6. '자치구'별 총 (고유) 운행 횟수 계산 ---
df_gu_runs_unique = df_joined.drop_duplicates(subset=['자치구', '노선번호'])
df_gu_total_runs = df_gu_runs_unique.groupby('자치구')['총운행횟수'].sum()

# --- 7. 최종 지표(Metric) 계산 ---
# '자치구'별 총 승하차 인원(5번)을 '자치구'별 총 운행 횟수(6번)로 나눔
df_metric = (df_gu_passengers['총승하차_인원'] / df_gu_total_runs).reset_index(name='회차당_총승객수')

# 8. 결과 정렬 (그래프 가독성을 위해)
df_metric_sorted = df_metric.sort_values(by='회차당_총승객수', ascending=False)

# --- 9. 그래프 생성 및 저장 ---
plt.figure(figsize=(15, 8)) # 25개 자치구를 표시하기 위해 가로를 길게
sns.barplot(
    data=df_metric_sorted,
    x='자치구',
    y='회차당_총승객수'
)

plt.title('자치구별 하루 회차당 총 승하차 인원 수', fontsize=16, pad=20)
plt.xlabel('자치구', fontsize=13)
plt.ylabel('회차당 총 승하차 인원 (명)', fontsize=13)
plt.xticks(rotation=90) # X축 레이블(자치구명) 90도 회전
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

# y축 레이블을 쉼표(,) 포맷으로 변경
current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels(['{:,.0f}'.format(x) for x in current_values])

# 그래프 저장
plot_filename = 'passengers_per_run_by_gu.png'
# bbox_inches='tight'는 x축 레이블이 잘리지 않게 함
plt.savefig(plot_filename, dpi=300, bbox_inches='tight')

print(f"'{plot_filename}' (자치구별 회차당 총승객수) 그래프가 저장되었습니다.")
print("\n--- 분석 결과 (상위 5) ---")
print(df_metric_sorted.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

# ==============================================================================
# 1. Matplotlib 한글 폰트 설정
# ==============================================================================
# Windows 환경 사용자는 'Malgun Gothic'이 기본으로 설정됩니다.
try:
    plt.rcParams["font.family"] = 'Malgun Gothic'
except:
    # 폰트 설정이 실패할 경우, 다른 기본 폰트를 시도하거나 경고를 출력합니다.
    pass
    
plt.rcParams["axes.unicode_minus"] = False # 마이너스 부호 깨짐 방지

print(f"Matplotlib 폰트 설정 완료: {plt.rcParams['font.family']}")


# ==============================================================================
# 2. 데이터 로드 및 전처리 (상관관계 분석을 위한 데이터 병합)
# ==============================================================================

# 파일 이름을 사용자의 환경에 맞게 로드합니다.
# 주의: 이 셀을 실행하기 전에 df_seoul_moving.csv와 SEOUL_BUS_STOP.csv 파일이 있어야 합니다.
df_moving = pd.read_csv("df_seoul_moving.csv")
df_bus_stop = pd.read_csv("SEOUL_BUS_STOP.csv", encoding='cp)

# 2.1. 유동인구 데이터 처리: '자치구'별 '이동인구(합)' 합계
df_moving_agg = df_moving.groupby('자치구')['이동인구(합)'].sum().reset_index()
df_moving_agg.rename(columns={'이동인구(합)': '유동인구_합계'}, inplace=True)

# 2.2. 버스 정류장 데이터 처리: '행정구명'별 고유 정류소(ARS-ID) 개수
df_bus_stop_agg = df_bus_stop.groupby('행정구명')['ARS-ID'].nunique().reset_index()
df_bus_stop_agg.rename(columns={'행정구명': '자치구', 'ARS-ID': '정류장_개수'}, inplace=True)

# 2.3. 데이터 병합
df_merged = pd.merge(df_moving_agg, df_bus_stop_agg, on='자치구', how='inner')


# ==============================================================================
# 3. 상관계수 계산
# ==============================================================================
correlation = df_merged['정류장_개수'].corr(df_merged['유동인구_합계'])
print(f"버스정류장 개수와 유동인구 합계의 피어슨 상관계수: {correlation:.4f}")


# ==============================================================================
# 4. 산점도 그래프 생성 (정류장 수=X축, 유동인구=Y축)
# ==============================================================================
plt.figure(figsize=(12, 10))

# 산점도 생성
plt.scatter(
    df_merged['정류장_개수'],  # X축: 정류장 개수
    df_merged['유동인구_합계'], # Y축: 유동인구 합계
    color='darkgreen',
    alpha=0.7,
    s=150 # 점 크기
)

# 점 위에 자치구 이름 레이블 (Annotation) 추가
for i, row in df_merged.iterrows():
    plt.annotate(
        row['자치구'],
        xy=(row['정류장_개수'], row['유동인구_합계']),
        xytext=(5, 5), # 점 옆으로 오프셋
        textcoords='offset points',
        ha='left',
        va='center',
        fontsize=9,
        bbox=dict(boxstyle="round,pad=0.2", fc="white", alpha=0.7, ec="none")
    )

# 라벨 및 제목 설정
plt.title(f'자치구별 유동인구와 버스정류장 개수의 상관관계 분석\n(상관계수: {correlation:.4f})', fontsize=16)
plt.xlabel('정류장 개수 (Number of Bus Stops)', fontsize=14)
plt.ylabel('유동인구 합계 (Sum of Floating Population)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.savefig('correlation_bus_stop_vs_population.png')
plt.show() # 주피터 환경에서 실행 결과 바로 확인
```
